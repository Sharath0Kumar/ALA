```python
class UpiId:
  def __init__(self, id , bank_id):
    self.my_id = id
    self.my_bank_id = bank_id

  def __repr__(self):
    return "upi {" + self.my_id + "@" + self.my_bank_id + "}"

  def __eq__(self , other):
    return self.my_id == other.my_id and self.my_bank_id == other.my_bank_id
```

```python
sk_upi_id is other_upi_id
```

```python
isinstance(sk_upi_id,UpiId)
```

```python
sk_upi_id = UpiId("6894574996" , "okhdfc")
```

```python
other_upi_id = UpiId("6894574996" , "okhdfc")
```

```python
sk_upi_id == other_upi_id
```

```python
isinstance(sk_upi_id , UpiId)
```

```python
isinstance(other_upi_id , UpiId)
```
